```python
import fastf1

session = fastf1.get_session(2024, "Bahrain", "R")
session.load()

results = session.results

results.head()
```

```python
results.columns
results[[
    "Abbreviation",
    "TeamName",
    "Position",
    "GridPosition",
    "Status"
]]
```

```python
results[
    ["Abbreviation", "GridPosition", "Position"] #Сравнение финишной позиции со стартовой
].sort_values("Position").head(10)
```

```python
results["PositionsGained"] = (
    results["GridPosition"] - results["Position"] 
) #Фьючер для подсчета позиций, которые гонщик выиграл или проиграл относительно старта
```

```python
results[
    ["Abbreviation", "GridPosition", "Position", "PositionsGained"]
].sort_values(
    "PositionsGained",
    ascending=False
)
```
