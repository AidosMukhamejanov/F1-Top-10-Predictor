```python
import fastf1
import pandas as pd
import os

CACHE_DIR = "cache"

fastf1.Cache.enable_cache(CACHE_DIR)

print("FastF1 cache enabled:", CACHE_DIR)
```

```python
session = fastf1.get_session(2022, 1, "R")
session.load()

results = session.results

print(results.shape)
display(results.head())
```

```python
import time

seasons = {
    2022: 22,
    2023: 22,
    2024: 24
}

all_races = []

for year, last_round in seasons.items():

    for round_number in range(1, last_round + 1):

        print(f"\n{'=' * 50}")
        print(f"Loading {year} Round {round_number}")
        print(f"{'=' * 50}")

        try:
            session = fastf1.get_session(year, round_number, "R")
            session.load()

            results = session.results.copy()

            results["Year"] = year
            results["Round"] = round_number

            all_races.append(results)

            print(f"SUCCESS | {year} Round {round_number} | {len(results)} drivers")

        except Exception as e:
            print(f"FAILED | {year} Round {round_number}")
            print(e)

        time.sleep(1)

print("\n" + "=" * 50)
print("DONE")
print("=" * 50)

print(f"Total races loaded: {len(all_races)}")
```

```python
full_df = pd.concat(all_races, ignore_index=True)

print("Shape:", full_df.shape)
print("Seasons:", sorted(full_df["Year"].unique()))
print("Races:", full_df[["Year", "Round"]].drop_duplicates().shape[0])

display(full_df.head())
```

```python
os.makedirs("data/processed", exist_ok=True)

output_path = "data/processed/f1_race_results_2022_2024.csv"

full_df.to_csv(output_path, index=False)

print(f"Saved: {output_path}")
print(f"Rows: {len(full_df)}")
print(f"Columns: {len(full_df.columns)}")
```
